### Feature Engineering

```python
import pandas as pd

df = pd.read_csv("/content/social_media_screentime_mental_health_2026_final.csv")
```

```python
# Fill missing values
df["avg_sleep_hours"] = df["avg_sleep_hours"].fillna(df["avg_sleep_hours"].mean())
df["gender"] = df["gender"].fillna(df["gender"].mode()[0])

# Remove unnecessary columns
df.drop("participant_id", axis=1, inplace=True)
df.drop("daily_notifications", axis=1, inplace=True)
df.drop("wellbeing_band", axis=1, inplace=True)
```

```python
print(df["night_time_use"].unique())
print(df["uses_screen_time_limits"].unique())
print(df["attempted_digital_detox"].unique())
```

```python
# Night time use score
night_score_map = {
    "Never": 0,
    "Sometimes": 4,
    "Often": 7,
    "Every night": 10
}

df["night_time_score"] = df["night_time_use"].map(night_score_map)

# Screen time limit score
screen_limit_map = {
    "Yes": 0,
    "No": 3
}

df["screen_limit_score"] = df["uses_screen_time_limits"].map(screen_limit_map)

# Digital detox score
detox_score_map = {
    "Yes, succeeded": 0,
    "Yes, failed": 1,
    "No": 2
}

df["digital_detox_score"] = df["attempted_digital_detox"].map(detox_score_map)
```

```python
df[[
    "night_time_use",
    "night_time_score",
    "uses_screen_time_limits",
    "screen_limit_score",
    "attempted_digital_detox",
    "digital_detox_score"
]].head(10)
```

```python
# Screen Time Score (0–25)

df["screen_time_score"] = (
    (
        df["daily_screen_hours"] - df["daily_screen_hours"].min()
    ) /
    (
        df["daily_screen_hours"].max() -
        df["daily_screen_hours"].min()
    )
) * 25

# Anxiety Score (0–15)

df["anxiety_score"] = (
    df["anxiety_score_0to27"] / 27
) * 15

# FOMO Score (0–15)

df["fomo_score"] = (
    (df["fomo_1to10"] - 1) / 9
) * 15


# Sleep Score (0–10)

df["sleep_score"] = (
    1 -
    (
        (df["avg_sleep_hours"] - df["avg_sleep_hours"].min()) /
        (
            df["avg_sleep_hours"].max() -
            df["avg_sleep_hours"].min()
        )
    )
) * 10


# Social Comparison Score (0–5)

df["social_comparison_score"] = (
    (df["social_comparison_1to10"] - 1) / 9
) * 5


# Physical Activity Score (0–5)

df["physical_activity_score"] = (
    1 -
    (
        df["physical_activity_days_per_week"] / 7
    )
) * 5


# First Check Score (0–10)

df["first_check_score"] = (
    1 -
    (
        df["minutes_to_first_check_after_waking"] /
        df["minutes_to_first_check_after_waking"].max()
    )
) * 10
```

```python
df[[
    "screen_time_score",
    "anxiety_score",
    "fomo_score",
    "sleep_score",
    "social_comparison_score",
    "physical_activity_score",
    "first_check_score"
]].head()
```

```python
# Calculate Total Addiction Score

df["Addiction_Score"] = (
    df["screen_time_score"] +
    df["anxiety_score"] +
    df["fomo_score"] +
    df["sleep_score"] +
    df["night_time_score"] +
    df["first_check_score"] +
    df["social_comparison_score"] +
    df["physical_activity_score"] +
    df["screen_limit_score"] +
    df["digital_detox_score"]
)
```

```python
df["Addiction_Score"].describe()
```

```python
# Create Addiction Level using quantiles

df["Addiction_Level"] = pd.qcut(
    df["Addiction_Score"],
    q=3,
    labels=["Low", "Moderate", "High"]
)
```

# Train-Test Split

```python
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

## Model Training

```python
#Import the models

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
```

```python
#Create the Models

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),

    "Decision Tree": DecisionTreeClassifier(random_state=42),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ),

    "KNN": KNeighborsClassifier(n_neighbors=5),

    "SVM": SVC(random_state=42),

    "Naive Bayes": GaussianNB()
}
```

```python
# Train all models

for model_name, model in models.items():

    print(f"Training {model_name}...")

    model.fit(X_train, y_train)

print("\nAll models trained successfully!")
```

## Model Evaluation

```python
#Import Evaluation Libraries

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

import pandas as pd
```

```python
#Evaluate All Models

# Store model results
results = []

for model_name, model in models.items():

    # Predict on testing data
    y_pred = model.predict(X_test)

    # Calculate accuracy
    accuracy = accuracy_score(y_test, y_pred)

    # Store the result
    results.append({
        "Model": model_name,
        "Accuracy": round(accuracy * 100, 2)
    })

# Create a results table
results_df = pd.DataFrame(results)

# Sort by accuracy
results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
)

results_df
```

## K-Means Clustering Module

```python
# K-Means imports

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import pandas as pd

# Copy only input features

X_cluster = X.copy()

print(X_cluster.shape)
```

```python
#Standardize the data

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_cluster)

print(X_scaled.shape)

#Find optimal number of clusters (Elbow Method)

inertia = []

for k in range(1, 10):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    inertia.append(kmeans.inertia_)


plt.figure(figsize=(8,5))

plt.plot(
    range(1,10),
    inertia,
    marker='o'
)

plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal K")

plt.show()

```

```python
#Apply K-means

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X_scaled)

print(cluster_labels[:10])

#Add cluster result to dataset

df["Cluster"] = cluster_labels

df.head()
```

```python
#Analyze each Cluster

cluster_analysis = df.groupby("Cluster").mean(numeric_only=True)

cluster_analysis

cluster_names = {
    0: "Healthy Users",
    1: "Moderate Risk Users",
    2: "High Risk Users"
}


df["Behaviour_Group"] = df["Cluster"].map(cluster_names)


df[["Cluster","Behaviour_Group"]].head()
```
